# Quartile Calculation and Export

This notebook reads `data/final_dataset.csv`, calculates the quartiles for key variables (Median House Price, Total Crime, Violence Against The Person, Total IMD, and Transport), and exports the resulting dataset to `quartiles.csv`.

```python
import pandas as pd
import numpy as np
```

```python
# 1. Read the dataset
df = pd.read_csv('data/final_dataset.csv')
print(f"Loaded dataset with {df.shape[0]} rows and {df.shape[1]} columns.")
```

```python
# 2. Calculate quartiles with string labels (e.g. 'Q1 (0.00 to 10.00)')
def calculate_quartile_with_labels(series):
    quantiles = series.quantile([0, 0.25, 0.5, 0.75, 1.0]).values
    labels = []
    for i in range(4):
        # Format nicely: Q1 (min to 25th), Q2 (25th to 50th), etc.
        labels.append(f"Q{i+1} ({quantiles[i]:.2f} to {quantiles[i+1]:.2f})")
    try:
        return pd.qcut(series, q=4, labels=labels)
    except ValueError:
        # Fallback to rank-based qcut if duplicate values prevent clean bin edges
        return pd.qcut(series.rank(method='first'), q=4, labels=labels)

df['Median House Price Quartile'] = calculate_quartile_with_labels(df['Median House Price 2022'])
df['Total Crime Quartile'] = calculate_quartile_with_labels(df['Total Crime 2022'])
df['Violence Against The Person Quartile'] = calculate_quartile_with_labels(df['Violence Against The Person'])
df['Total IMD Quartile'] = calculate_quartile_with_labels(df['IMD Score'])
df['Transport Quartile'] = calculate_quartile_with_labels(df['Average PTAI 2015'])

print("Quartiles successfully calculated with range labels.")
```

```python
# 3. Select columns and save to CSV
output_cols = [
    'LSOA code', 
    'LSOA name', 
    'Local authority name',
    'Median House Price Quartile',
    'Total Crime Quartile',
    'Violence Against The Person Quartile',
    'Total IMD Quartile',
    'Transport Quartile'
]

quartiles_df = df[output_cols]
quartiles_df.to_csv('quartiles.csv', index=False)
print("Saved quartiles to quartiles.csv")
quartiles_df.head()
```
